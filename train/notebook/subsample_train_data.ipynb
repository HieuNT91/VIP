```python
import json 
import random
from datasets import load_dataset
```

```python
file_path = "/storage2/xxx/VIP/clone_space/verl_forked/data/regression_data/allo_grpo_4e/per_question_statistics_latest.json"
# result_path = json.loads(file_path)
with open(file_path, "r") as f:
    data = json.load(f)
```

```python
keep_keys = []
for key, value in data.items():
    # print(f"Question ID: {key}, Statistics: {value}")
    # break
    if value["mean_acc_per_epoch"][0] == 0:
        continue
    if value["std_acc_per_epoch"][0] == 0:
        continue
    keep_keys.append(key)
print(len(keep_keys))
```

```python
subsample_size = 6000
subsampled_keep_keys = random.sample(keep_keys, subsample_size)

```

```python
dataset_path = "/storage2/xxx/VIP/data/vip-dapo-math-17k.parquet"
data = load_dataset("parquet", data_files=dataset_path)
```

```python
subsampled_data = data["train"].filter(lambda example: example["extra_info"]['index'] in subsampled_keep_keys)
subsampled_data.to_parquet("/storage2/xxx/VIP/data/vip-dapo-math-6k-subsampled.parquet")
```
